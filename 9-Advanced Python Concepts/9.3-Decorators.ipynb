```python
## closures

def main_welcome(msg):
    
    def sub_welcome_method():
        print("welcome to the advanced course")
        print(msg)
        print("pls learn these concepts")
    return sub_welcome_method()
```

```python
main_welcome("wlc")
```

```python
## Decorator
def my_decorator(func):
    
    def wrapper():
        print("welcome to the advanced course")
        func()
        print("pls learn these concepts")
    return wrapper
```

```python
@my_decorator
def say_hlo():
    print("hlo")
```

```python
say_hlo()
```
