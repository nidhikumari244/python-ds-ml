```python
##EX__ cal the sum of n natural no using a while and for loop

n= 55
sum = 0
count=1

while count <=n:
    sum = sum+count
    count=count+1
print("sum of 1st 10 natural no.",sum)
```

```python
## prime no b/w 1 and 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)

```
