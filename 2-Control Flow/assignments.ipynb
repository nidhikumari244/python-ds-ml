### CONDITIONAL STATEMENT ASSIGNMENT


```python
## 1 if 
num = int(input("enter a no."))
if num>0:
    print(num,"is positive")
```

```python
##2 if else
num = int(input("enter a no."))
if num>0:
    print(num,"is positive")
else:
    print(num,"is negative")
```

```python
##3 if elif else

num = int(input("enter a number"))
if num > 0:
    print(num,"is positive no.")
elif num < 0:
    print(num,"is negative no.")
else:
    print("zero")
```

```python
##4 nested if

num = int(input("enter a no."))
if num>0:
    if num%2==0:
        print(num,"is +ve no. and  even")
    else:
        print(num,"is +ve no.and odd")
else:
    print(num,"is -ve")
```

```python
## 8 
total = 0
while True:
    num=float(input("enter the no.(0 to stop): "))
    if num == 0:
        break
    total += num
print(f"the sum of all no is{total} ")
```

```python
# 11.
num = int(input("enter a no."))
for i in range(1,num+1):
    if i%2==0:
        print(i,"is even")

```

```python
# factorial of no. 
num = int(input("enter a no."))
factorial = 1
i = 1
while i<=num:
    factorial*=i
    i+=1
print(f"factorial of {num} is {factorial}")
```

```python
#13

num=int(input("enter a no."))
sum_of_digits=0
while num>0:
    digits=num%10
    sum_of_digits+=digits
    num=num//10
print(f"sum of digits is {sum_of_digits}")
```
