```python
students={
    "student1": {"name":"k","age":222},
    "student2": {"name": "r", "age":2222}
}
print(students)
```

```python
## iterating over nested dictionary

print(students["student1"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
n=[11,22,22,4,4,45,6,6,6,6,6,6,6,78,78,6,11]
frequency={}
for number in n:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## merge 2 dic. into 1
dic1={"a":1,"b":2}
dic2={"b":34,"c":4}
merged_dic={**dic1,**dic2}
print(merged_dic)
```
