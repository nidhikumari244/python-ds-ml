```python
# LIST METHOD

import random
random_number= [random.randint(1,20) for _ in range(15)]
print(f"original numbers: {random_number}")

sorted_number= sorted(random_number)
print(f"sorted in ascending no.:{sorted_number}")

sorted_number_desc= sorted(random_number, reverse=True)
print(f"sorted in descending order:{sorted_number_desc}")

unique_number= list(set(random_number))
print(f"list with duplicate remvoed{ unique_number}")
```
